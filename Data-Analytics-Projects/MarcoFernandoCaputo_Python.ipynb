# MeditActive - Sistema di Gestione degli Elementi 

Questo progetto sviluppa un piccolo sistema in Python per creare e gestire le funzionalità pricipali degli utenti MeditActive.

Gli elementi rappresentano obiettivi o attivita personali, per esempio sessioni di meditazione, attività fisica, pratiche di crescita personale o iniziative sostenibili.

Funzionalità:
- registrazione di nuovi elementi;
- visualizzazione completa della collezione;
- ricerca semplice per uno o più attributi;
- statistiche sugli elementi;
- filtraggio con condizioni multiple.

## Struttura dei dati

La collezione viene salvata in una lista chiamata `elementi`. Ogni elemento è un dizionario Python con questi attributi:

- `nome`: nome dell'obiettivo o attivita;
- `tipo`: categoria dell'elemento;
- `frequenza_attivita_fisica`: livelli come `bassa`, `media` o `alta`;
- `durata_minuti`: durata stimata dell'attività;
- `obiettivo_temporale`: ad esempio `giornaliero`, `mensile` o `annuale`;
- `monete`: monete virtuali associate al completamento;
- `data_creazione`: data di registrazione;
- `note`: informazioni aggiuntive.

```python
# importa date per lavorare con classe data e counter per il conteggio "automatico"

from datetime import date
from collections import Counter

# Lista principale che contiene tutti gli elementi registrati.
elementi = []
```

## Funzioni principali

Le funzioni seguenti permettono di registrare, visualizzare, cercare, filtrare e analizzare gli elementi presenti nella collezione.

```python
def normalizza_testo(valore):
    """Converte un valore in testo minuscolo e senza spazi esterni."""
    return str(valore).strip().lower()


def crea_elemento(nome, tipo, frequenza_attivita_fisica, durata_minuti,
                  obiettivo_temporale, monete, note=""):
    """Crea e restituisce un dizionario che rappresenta un elemento MeditActive."""
    elemento = {
        "nome": nome.strip(),
        "tipo": tipo.strip().lower(),
        "frequenza_attivita_fisica": frequenza_attivita_fisica.strip().lower(),
        "durata_minuti": int(durata_minuti),
        "obiettivo_temporale": obiettivo_temporale.strip().lower(),
        "monete": int(monete),
        "data_creazione": date.today().isoformat(),
        "note": note.strip()
    }
    return elemento


def registra_elemento(collezione, nome, tipo, frequenza_attivita_fisica,
                      durata_minuti, obiettivo_temporale, monete, note=""):
    """Aggiunge un nuovo elemento alla lista e lo restituisce, evitando duplicati."""
    nome_normalizzato = normalizza_testo(nome)
    tipo_normalizzato = normalizza_testo(tipo)

    for elemento in collezione:
        if (normalizza_testo(elemento["nome"]) == nome_normalizzato and
                normalizza_testo(elemento["tipo"]) == tipo_normalizzato):
            print(
                f"Elemento '{str(nome).strip()}' di tipo '{tipo_normalizzato}' gia presente. "
                "Registrazione annullata."
            )
            return None

    nuovo_elemento = crea_elemento(
        nome,
        tipo,
        frequenza_attivita_fisica,
        durata_minuti,
        obiettivo_temporale,
        monete,
        note
    )
    collezione.append(nuovo_elemento)
    print(f"Elemento '{nuovo_elemento['nome']}' registrato correttamente.")
    return nuovo_elemento


def stampa_elemento(elemento, numero=None):
    """Stampa un elemento in modo leggibile."""
    intestazione = f"Elemento {numero}" if numero is not None else "Elemento"
    print(intestazione)
    print("-" * len(intestazione))
    print(f"Nome: {elemento['nome']}")
    print(f"Tipo: {elemento['tipo']}")
    print(f"Frequenza attivita fisica: {elemento['frequenza_attivita_fisica']}")
    print(f"Durata: {elemento['durata_minuti']} minuti")
    print(f"Obiettivo temporale: {elemento['obiettivo_temporale']}")
    print(f"Monete: {elemento['monete']}")
    print(f"Data creazione: {elemento['data_creazione']}")
    if elemento['note']:
        print(f"Note: {elemento['note']}")
    print()


def visualizza_elementi(collezione):
    """Visualizza tutti gli elementi presenti nella lista."""
    if len(collezione) == 0:
        print("Non ci sono elementi registrati.")
        return

    print(f"Elementi registrati: {len(collezione)}\n")
    for indice, elemento in enumerate(collezione, start=1):
        stampa_elemento(elemento, indice)
```

## Ricerca semplice

La funzione `cerca_elementi` cerca gli elementi confrontando uno o piu attributi. Per i testi viene usato un confronto non sensibile alle maiuscole; per i numeri viene usato il valore esatto.

```python
def cerca_elementi(collezione, **criteri):
    """Cerca elementi che rispettano tutti i criteri specificati."""
    risultati = []

    for elemento in collezione:
        corrisponde = True

        for attributo, valore_cercato in criteri.items():
            if attributo not in elemento:
                corrisponde = False
            else:
                valore_elemento = elemento[attributo]

                if isinstance(valore_elemento, str):
                    if normalizza_testo(valore_elemento) != normalizza_testo(valore_cercato):
                        corrisponde = False
                else:
                    if valore_elemento != valore_cercato:
                        corrisponde = False

        if corrisponde:
            risultati.append(elemento)

    return risultati
```

## Filtraggio avanzato

Il filtro avanzato permette di combinare piu condizioni nello stesso momento. Per esempio: trovare tutti gli elementi di tipo `meditazione`, con durata massima di 20 minuti e obiettivo `giornaliero`.

```python
def filtra_elementi_avanzato(collezione, tipo=None, frequenza_attivita_fisica=None,
                             durata_min=None, durata_max=None, monete_min=None,
                             monete_max=None, obiettivo_temporale=None,
                             testo_nome=None):
    """Filtra gli elementi usando condizioni multiple opzionali."""
    risultati = []

    for elemento in collezione:
        valido = True

        if tipo is not None and normalizza_testo(elemento["tipo"]) != normalizza_testo(tipo):
            valido = False

        if frequenza_attivita_fisica is not None:
            if normalizza_testo(elemento["frequenza_attivita_fisica"]) != normalizza_testo(frequenza_attivita_fisica):
                valido = False

        if durata_min is not None and elemento["durata_minuti"] < durata_min:
            valido = False

        if durata_max is not None and elemento["durata_minuti"] > durata_max:
            valido = False

        if monete_min is not None and elemento["monete"] < monete_min:
            valido = False

        if monete_max is not None and elemento["monete"] > monete_max:
            valido = False

        if obiettivo_temporale is not None:
            if normalizza_testo(elemento["obiettivo_temporale"]) != normalizza_testo(obiettivo_temporale):
                valido = False

        if testo_nome is not None:
            if normalizza_testo(testo_nome) not in normalizza_testo(elemento["nome"]):
                valido = False

        if valido:
            risultati.append(elemento)

    return risultati
```

## Statistiche sugli elementi

La funzione `statistiche_elementi` produce una panoramica della collezione: totale elementi, distribuzione per tipo, frequenza dell'attivita fisica, obiettivi temporali, totale monete e durata media.

```python
def statistiche_elementi(collezione):
    """Genera statistiche utili sugli elementi presenti nella collezione."""
    if len(collezione) == 0:
        print("Non ci sono elementi da analizzare.")
        return {}

    tipi = []
    frequenze = []
    obiettivi = []
    totale_monete = 0
    totale_durata = 0

    for elemento in collezione:
        tipi.append(elemento["tipo"])
        frequenze.append(elemento["frequenza_attivita_fisica"])
        obiettivi.append(elemento["obiettivo_temporale"])
        totale_monete += elemento["monete"]
        totale_durata += elemento["durata_minuti"]

    statistiche = {
        "totale_elementi": len(collezione),
        "distribuzione_per_tipo": dict(Counter(tipi)),
        "distribuzione_frequenza_attivita_fisica": dict(Counter(frequenze)),
        "distribuzione_obiettivi_temporali": dict(Counter(obiettivi)),
        "totale_monete": totale_monete,
        "durata_media_minuti": round(totale_durata / len(collezione), 2),
        "tipo_piu_frequente": Counter(tipi).most_common(1)[0][0],
        "elemento_piu_remunerativo": max(collezione, key=lambda elemento: elemento["monete"])["nome"]
    }

    print("Statistiche MeditActive")
    print("-" * 24)
    print(f"Totale elementi: {statistiche['totale_elementi']}")
    print(f"Distribuzione per tipo: {statistiche['distribuzione_per_tipo']}")
    print(f"Frequenza attivita fisica: {statistiche['distribuzione_frequenza_attivita_fisica']}")
    print(f"Obiettivi temporali: {statistiche['distribuzione_obiettivi_temporali']}")
    print(f"Totale monete ottenibili: {statistiche['totale_monete']}")
    print(f"Durata media: {statistiche['durata_media_minuti']} minuti")
    print(f"Tipo piu frequente: {statistiche['tipo_piu_frequente']}")
    print(f"Elemento piu remunerativo: {statistiche['elemento_piu_remunerativo']}")

    return statistiche
```

## Menù interattivo opzionale

La funzione seguente permette di usare il sistema tramite input da tastiera. Nel notebook non viene avviata automaticamente, cosi l'esecuzione completa delle celle non resta bloccata in attesa di input.

Per avviarla, eseguire in una nuova cella: `menu_sge()`

```python
def menu_sge():
    """Menu testuale opzionale per gestire la collezione da Jupyter."""
    while True:
        print("\nMeditActive SGE")
        print("1. Registra nuovo elemento")
        print("2. Visualizza tutti gli elementi")
        print("3. Cerca per tipo")
        print("4. Filtra per tipo e durata massima")
        print("5. Mostra statistiche")
        print("0. Esci")

        scelta = input("Scegli un'opzione: ")

        if scelta == "1":
            nome = input("Nome: ")
            tipo = input("Tipo: ")
            frequenza = input("Frequenza attivita fisica (bassa/media/alta): ")
            try:
                durata = int(input("Durata in minuti: "))
            except ValueError:
                print("La durata deve essere un numero intero. Riprova.")
                continue

            obiettivo = input("Obiettivo temporale (giornaliero/mensile/annuale): ")

            try:
                monete = int(input("Monete: "))
            except ValueError:
                print("Le monete devono essere un numero intero. Riprova.")
                continue

            note = input("Note: ")

            registra_elemento(elementi, nome, tipo, frequenza, durata, obiettivo, monete, note)

        elif scelta == "2":
            visualizza_elementi(elementi)

        elif scelta == "3":
            tipo = input("Tipo da cercare: ")
            risultati = cerca_elementi(elementi, tipo=tipo)
            visualizza_elementi(risultati)

        elif scelta == "4":
            tipo = input("Tipo da cercare: ")
            try:
                durata_max = int(input("Durata massima in minuti: "))
            except ValueError:
                print("La durata massima deve essere un numero intero. Riprova.")
                continue

            risultati = filtra_elementi_avanzato(elementi, tipo=tipo, durata_max=durata_max)
            visualizza_elementi(risultati)

        elif scelta == "5":
            statistiche_elementi(elementi)

        elif scelta == "0":
            print("Chiusura del sistema MeditActive SGE.")
            break

        else:
            print("Scelta non valida. Riprova.")
```
